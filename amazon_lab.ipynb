```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\rkham\OneDrive\Desktop\amz_uk_price_prediction_dataset.csv\amz_uk_price_prediction_dataset.csv")

df.head()
```

```python
df.info()
df[['category', 'price', 'stars']].isnull().sum()
```

```python
category_counts = df['category'].value_counts()

print(category_counts.head(5))
```

```python
top_10 = category_counts.head(10)

plt.figure(figsize=(12, 5))
sns.barplot(x=top_10.values, y=top_10.index)

plt.title("Top 10 Product Categories")
plt.xlabel("Number of Products")
plt.ylabel("Category")
plt.show()
```

```python
top_5 = category_counts.head(5)

plt.figure(figsize=(7, 7))
plt.pie(top_5.values, labels=top_5.index, autopct='%1.1f%%')
plt.title("Proportion of Top 5 Categories")
plt.show()
```

```python
price = df['price']

print("Mean:", price.mean())
print("Median:", price.median())
print("Mode:", price.mode()[0])
```

```python
print("Variance:", price.var())
print("Standard deviation:", price.std())
print("Range:", price.max() - price.min())
print("IQR:", price.quantile(0.75) - price.quantile(0.25))
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(price[0:1000], bins=1000)

plt.title("Distribution of Product Prices")
plt.xlabel("Price (£)")
plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.boxplot(x=price[0:1000])

plt.title("Product Price Box Plot")
plt.show()
```

```python
rating = df['stars']

print("Mean:", rating.mean())
print("Median:", rating.median())
print("Mode:", rating.mode()[0])
```

```python
print("Variance:", rating.var())
print("Standard deviation:", rating.std())
print("IQR:", rating.quantile(0.75) - rating.quantile(0.25))
```

```python
print("Skewness:", rating.skew())
print("Kurtosis:", rating.kurt())
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(rating, bins=100)

plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.show()
```
